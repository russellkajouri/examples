```python
import matplotlib.pyplot as plt
import numpy as np

scale = 0.5
plt.style.use('seaborn-v0_8-bright')
plt.rcParams.update({
    'font.size': 14*scale,
    'lines.linewidth': 2 * scale,
    'axes.linewidth':  1 * scale,
    'xtick.major.size':  6 * scale,
    'ytick.major.size':  6 * scale,
    'xtick.minor.size':  3 * scale,
    'ytick.minor.size':  3 * scale,
    'xtick.major.width': 1 * scale,
    'ytick.major.width': 1 * scale,
    'xtick.minor.width': 1 * scale,
    'ytick.minor.width': 1 * scale,
    'text.usetex' : True,
    'mathtext.fontset' : 'cm'
})
```

```python
f1 = '../modified1/co2_stats.dat'
f2 = '../modified2/co2_stats.dat'

f3 = '../modified1/acn_stats.dat'
f4 = '../modified2/acn_stats.dat'

d1 = np.loadtxt(f1)
d2 = np.loadtxt(f2)
d3 = np.loadtxt(f3)
d4 = np.loadtxt(f4)
```

```python
def take_data(d1,d2,d3,d4):
    assert d1.shape == d2.shape == d3.shape == d4.shape, 'no compatible'

    rows, clmns = d1.shape
    gcmcCycles = np.arange(0, rows)

    ylabels = [
        'insertion attempts',
        'insertion success',
        'deletion attempts',
        'deletion success',
        r'$\langle N \rangle$'
    ]


    fig = plt.figure(figsize=(5, (clmns-1) * 2.5), dpi=300)
    gs = fig.add_gridspec(clmns, 2, hspace=0.1)
    axes = np.array([[fig.add_subplot(gs[r, 0]), fig.add_subplot(gs[r, 1])] for r in range(clmns-2)])
    
    for c in range(1, clmns-1):
        axes[c-1, 0].plot(gcmcCycles, d1.T[c], ls='solid', label=f1)
        axes[c-1, 0].plot(gcmcCycles, d2.T[c], ls='dashed', label=f2)

        axes[c-1, 1].plot(gcmcCycles, d3.T[c], ls='solid', label=f1)
        axes[c-1, 1].plot(gcmcCycles, d4.T[c], ls='dashed', label=f2)

        #axes[c-1].plot(gcmcCycles, d1.T[c], marker='o', ms=2, lw=0, label=f1)
        #axes[c-1].plot(gcmcCycles, d2.T[c], marker='x', ms=2, lw=0, label=f2)

        axes[c-1, 0].set_ylabel(ylabels[c-1])
        axes[c-1, 0].tick_params(labelbottom=False, direction='in')
        axes[c-1, 1].tick_params(labelbottom=False, direction='in')
    else:
        fig.align_ylabels()
        for n in range(2):
            axes[c-1, n].set_xlabel('GCMC Cycles')
            axes[c-1, n].tick_params(labelbottom=True)
            axes[c-1, n].legend(loc=2)
    plt.show()
```

```python
take_data(d1,d2,d3,d4)
```

```python
f1 = '../modified1/thermal_statistics.dat'
f2 = '../modified2/thermal_statistics.dat'

d1 = np.loadtxt(f1)
d2 = np.loadtxt(f2)
```

```python
def take_data(d1,d2):
    assert d1.shape == d2.shape, 'no compatible'
    rows, clmns = d1.shape
    gcmcCycles = np.arange(0, rows)

    ylabels = [
        'TEMP',
        'TEMP/compute',
        'Pressures',
        'Kinetic Energy',
        'Potenital Energy',
        'Total Energy'
    ]


    fig = plt.figure(figsize=(2.5, (clmns-1) * 2.5), dpi=300)
    gs = fig.add_gridspec(clmns, 1, hspace=0.1)
    axes = np.array([fig.add_subplot(gs[r, 0]) for r in range(clmns-2)])
    
    for c in range(1, clmns-1):
        axes[c-1].plot(gcmcCycles, d1.T[c], marker='o', ms=2, lw=0, label=f1)
        axes[c-1].plot(gcmcCycles, d2.T[c], marker='x', ms=2, lw=0, label=f2)

        axes[c-1].set_ylabel(ylabels[c-1])
        axes[c-1].tick_params(labelbottom=False, direction='in')
    else:
        fig.align_ylabels()
        axes[c-1].set_xlabel('GCMC Cycles')
        axes[c-1].tick_params(labelbottom=True)
        axes[c-1].legend(loc=2)
    plt.show()
```

```python
take_data(d1,d2)
```
